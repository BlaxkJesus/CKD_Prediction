```python
# IMPORT LIBRARIES
import pandas as pd
import numpy as np
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import pickle
```

```python
# LOAD DATASET
df = pd.read_csv("ckd dile.csv")
```

```python
# Clean column names
df.columns = df.columns.str.strip().str.lower()
```

```python
# DROP ID COLUMN
if 'id' in df.columns:
    df = df.drop('id', axis=1)
```

```python
# HANDLE MISSING VALUES
df.replace("?", np.nan, inplace=True)
```

```python
df = df.apply(pd.to_numeric, errors='coerce')
```

```python
# FINAL CLEAN (REMOVE ANY LEFTOVER NaN)
df = df.fillna(df.mean(numeric_only=True))
```

```python
# SPLIT FEATURES & LABEL
# =========================
X = df.drop("class", axis=1)
y = df["class"]
```

```python
# SCALE DATA
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Save scaler
pickle.dump(scaler, open("scaler.pkl", "wb"))
```

```python
# TRAIN TEST SPLIT
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

```python
# PREPARE CNN INPUT
X_train_cnn = X_train.reshape(-1, X_train.shape[1], 1)
X_test_cnn = X_test.reshape(-1, X_test.shape[1], 1)
```

```python
# BUILD MODEL
from tensorflow.keras.layers import Input, Dense, Conv1D, Flatten, Concatenate, Dropout
from tensorflow.keras.models import Model
```

```python
# CNN branch
cnn_input = Input(shape=(X_train.shape[1], 1))
x = Conv1D(32, 2, activation='relu')(cnn_input)
x = Conv1D(64, 2, activation='relu')(x)
x = Flatten()(x)
cnn_output = Dense(64, activation='relu')(x)
```

```python

# Dense (Transformer-like) branch
transformer_input = Input(shape=(X_train.shape[1],))
y_branch = Dense(64, activation='relu')(transformer_input)
y_branch = Dense(32, activation='relu')(y_branch)
```

```python
# Combine
combined = Concatenate()([cnn_output, y_branch])
z = Dense(64, activation='relu')(combined)
z = Dropout(0.3)(z)
output = Dense(1, activation='sigmoid')(z)
```

```python
# Model
model = Model(inputs=[cnn_input, transformer_input], outputs=output)
```

```python

# Compile
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# TRAIN MODEL
history = model.fit(
    [X_train_cnn, X_train],
    y_train,
    epochs=50,
    batch_size=8,
    validation_split=0.1
)
```

```python
# EVALUATE
loss, acc = model.evaluate([X_test_cnn, X_test], y_test)
print(f"Test Accuracy: {acc * 100:.2f}%")
```

```python
# SAVE MODEL
model.save("ckd_model.h5")
print("Model and scaler saved successfully!")
```
